# 第1回：IIIFで画像の範囲と大きさを変える

確認問題まで上から順にセルを実行してください（Shift+Enter）。値を変えたら、そのセルから下を再実行します。

確認問題の後の追加課題は授業外・任意です。予測とコードの変更は自分で行い、エラーが出たらAIに説明を求められます。

資料：[画本虫撰](https://codh.rois.ac.jp/pmjt/book/200014778/)（日本古典籍データセット、国文研所蔵、CODH配信、[CC BY-SA 4.0](https://creativecommons.org/licenses/by-sa/4.0/)）

## 1. 出典と利用条件を確かめる

IIIFのマニフェストには、資料の題名・提供元・利用条件が書かれています。

````python
import requests
from PIL import Image

manifest = requests.get("https://codh.rois.ac.jp/pmjt/book/200014778/manifest.json").json()
print(manifest["label"])
print(manifest["attribution"])
print(manifest["license"])
````

## 2. 見開き全体を取得する

`koma`はコマ番号です。URLの`full`は範囲（見開き全体）、`1000,`は大きさ（幅1000ピクセル、高さは縦横比に合わせる）です。

````python
koma = 5
url = f"https://codh.rois.ac.jp/pmjt/iiif/200014778/200014778_{koma:05d}.tif/full/1000,/0/default.jpg"
!curl -fso page.jpg {url}
print(url)
Image.open("page.jpg")
````

## 3. 範囲と大きさを変える

`region`は切り出す範囲（左,上,幅,高さ）で、原画像のピクセルで指定します。実行する前に、この範囲で見えなくなる情報を予測してください。

````python
region = "842,365,3875,3061"
size = "1200,"
url = f"https://codh.rois.ac.jp/pmjt/iiif/200014778/200014778_{koma:05d}.tif/{region}/{size}/0/default.jpg"
!curl -fso crop.jpg {url}
print(url)
Image.open("crop.jpg")
````

表示されたURLを開くと、誰でも同じ範囲と大きさの画像を取得できます。範囲を一か所変えて、絵と和歌の位置関係が分かるか比べてください。`size`を半分にした場合は、文字の形を確かめてください。

## 4. 別のコマで試す

`koma`を変えて2.から実行し直すと、同じ資料の別のページを取得できます。

## 確認問題

コマ6の左半分を幅500ピクセルで取得し、`answer.jpg`に保存してください。原画像は幅5616・高さ3744ピクセルです。下のセルの値を変えて実行し、次のセルで答えを確かめてください。

````python
koma = 5
region = "full"
size = "1000,"
url = f"https://codh.rois.ac.jp/pmjt/iiif/200014778/200014778_{koma:05d}.tif/{region}/{size}/0/default.jpg"
!curl -fso answer.jpg {url}
Image.open("answer.jpg")
````

````python
#@title 答えを確かめる
from PIL import ImageChops, ImageStat

!curl -fso expected.jpg "https://codh.rois.ac.jp/pmjt/iiif/200014778/200014778_00006.tif/0,0,2808,3744/500,/0/default.jpg"
answer, expected = Image.open("answer.jpg").convert("L"), Image.open("expected.jpg").convert("L")
if answer.size != expected.size:
    print(f"結果：不合格（画像の幅・高さが{answer.size}です。幅を500にしてください）")
elif ImageStat.Stat(ImageChops.difference(answer, expected)).mean[0] > 10:
    print("結果：不合格（大きさは合っています。コマと範囲を確かめてください）")
else:
    print("結果：合格")
````

## 授業外・任意：右半分の範囲を計算する

コマ6の左半分と右半分で、左端・上端・幅・高さのうちどの値が変わりますか。予測を書いてください。下のセルは左半分を取得するコードを折りたたんでいます。確認問題に合格してから、セルをダブルクリックしてコードを開き、`left`を書き換えてください。

`//`は整数で割った商を返します。確認問題で取得した左半分と見比べ、二枚が中央でつながるか確かめてください。

````python
#@title 追加課題：左半分を取得するコード（確認問題の後に開く）
width = 5616
height = 3744
half = width // 2
left = 0
koma = 6
size = "500,"
region = f"{left},0,{half},{height}"
url = f"https://codh.rois.ac.jp/pmjt/iiif/200014778/200014778_{koma:05d}.tif/{region}/{size}/0/default.jpg"
print(region)
!curl -fso half.jpg {url}
Image.open("half.jpg")
````
