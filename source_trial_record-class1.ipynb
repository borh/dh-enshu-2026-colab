# 第1回：IIIFで画像の範囲と大きさを変える

上から順にセルを実行してください（Shift+Enter）。値を変えたら、そのセルから下を再実行します。

資料：[官報 1902年8月20日](https://dl.ndl.go.jp/pid/2949042/1/5)（国立国会図書館デジタルコレクション）

## 1. 出典と利用条件を確かめる

IIIFのマニフェストには、資料の題名・提供元・利用条件が書かれています。

````python
import requests
from PIL import Image

manifest = requests.get("https://dl.ndl.go.jp/api/iiif/2949042/manifest.json").json()
print(manifest["label"])
print(manifest["attribution"])
print(manifest["license"])
````

## 2. ページ全体を取得する

`koma`はコマ番号です。URLの`full`は範囲（ページ全体）、`1000,`は大きさ（幅1000ピクセル、高さは縦横比に合わせる）です。

````python
koma = 5
url = f"https://dl.ndl.go.jp/api/iiif/2949042/R{koma:07d}/full/1000,/0/default.jpg"
!curl -fso page.jpg {url}
print(url)
Image.open("page.jpg")
````

## 3. 範囲と大きさを変える

`region`は切り出す範囲（左,上,幅,高さ）で、原画像のピクセルで指定します。実行する前に、この範囲で見えなくなる情報を予測してください。

````python
region = "2257,334,1059,1811"
size = "760,"
url = f"https://dl.ndl.go.jp/api/iiif/2949042/R{koma:07d}/{region}/{size}/0/default.jpg"
!curl -fso crop.jpg {url}
print(url)
Image.open("crop.jpg")
````

表示されたURLを開くと、誰でも同じ範囲と大きさの画像を取得できます。範囲を一か所変えて、報告の見出しや隣の行との関係が分かるか比べてください。`size`を半分にした場合は、文字の形を確かめてください。

## 4. 別のコマで試す

`koma`を変えて2.から実行し直すと、同じ資料の別のページを取得できます。

## 確認問題

コマ6の左半分を幅500ピクセルで取得し、`answer.jpg`に保存してください。原画像は幅6688・高さ4320ピクセルです。下のセルの値を変えて実行し、次のセルで答えを確かめてください。

````python
koma = 5
region = "full"
size = "1000,"
url = f"https://dl.ndl.go.jp/api/iiif/2949042/R{koma:07d}/{region}/{size}/0/default.jpg"
!curl -fso answer.jpg {url}
Image.open("answer.jpg")
````

````python
#@title 答えを確かめる
from PIL import ImageChops, ImageStat

!curl -fso expected.jpg "https://dl.ndl.go.jp/api/iiif/2949042/R0000006/0,0,3344,4320/500,/0/default.jpg"
answer, expected = Image.open("answer.jpg").convert("L"), Image.open("expected.jpg").convert("L")
if answer.size != expected.size:
    print(f"結果：不合格（画像の幅・高さが{answer.size}です。幅を500にしてください）")
elif ImageStat.Stat(ImageChops.difference(answer, expected)).mean[0] > 10:
    print("結果：不合格（大きさは合っています。コマと範囲を確かめてください）")
else:
    print("結果：合格")
````
